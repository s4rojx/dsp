```python
import pandas as pd

df.drop_duplicates(inplace=True)
print(df)

df1 = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Name': ['John', 'Anna', 'Peter', 'Linda'],
    'Age': [28, 22, 35, 32]
})

df2 = pd.DataFrame({
    'ID': [3, 4, 5, 6],
    'City': ['New York', 'Berlin', 'Paris', 'Tokyo'],
    'Salary': [70000, 80000, 50000, 60000]
})

inner_merge = pd.merge(df1, df2, on='ID', how='inner')
print("\nInner Merge:")
print(inner_merge)
```
